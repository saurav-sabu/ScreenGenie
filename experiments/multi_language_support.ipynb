```python
from langchain_google_genai import GoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_community.document_loaders import PyPDFLoader
import pymupdf4llm

from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
md_text = pymupdf4llm.to_markdown("data/Saurav_Sabu_Resume.pdf")
md_text
```

```python
translate_resume_prompt = """
Act as a professional resume translator specializing in job application documents. Translate the given resume into the target language while preserving the structure, tone, and professional terminology.

### Input:
- Resume Text: {resume_text}
- Target Language: {target_language}

### Output:
Provide the translated resume as plain text while maintaining the original formatting and ensuring professional language standards. If you encounter any ambiguous terms, note them in the comments.

### Example Output Format:
{{
    "TranslatedResume": "Translated resume content here...",
    "Comments": "Notes about ambiguous terms or context."
}}

"""
```

```python
prompt = ChatPromptTemplate.from_template(translate_resume_prompt)
```

```python
model = GoogleGenerativeAI(model="gemini-1.5-pro-latest")
chain = prompt | model | StrOutputParser()
```

```python
resume_text = md_text
target_language = "hindi"
```

```python
result = chain.invoke({"resume_text": resume_text, "target_language": target_language})
print(result)
```
